```python
import pandas as pd
```

```python
df = pd.read_csv("sales_data_sample.csv", encoding="latin1")
```

```python
df["CONTACTFIRSTNAME"] = df["CONTACTFIRSTNAME"].str.title()
```

```python
groups = df.groupby(["CONTACTFIRSTNAME", "CONTACTLASTNAME"])
groups["SALES"].agg(["sum", "mean", "max", "min", "count"])
```

```python
Munich = df["CITY"]=="Munich"
df.loc[Munich, "CONTACTFIRSTNAME"] = df.loc[Munich, "CONTACTFIRSTNAME"].str.title()
df.loc[Munich, "CONTACTLASTNAME"] = df.loc[Munich, "CONTACTLASTNAME"].str.title()
```

```python
df.groupby("YEAR_ID")["SALES"].sum()
```

```python
df.groupby("COUNTRY")["SALES"].sum().sort_values(ascending=False)
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
orders_per_year = df.groupby('YEAR_ID')['ORDERNUMBER'].nunique().sort_index()
plt.bar(orders_per_year.index.astype(str), orders_per_year.values, color='blue', edgecolor='black')
plt.title('Illere gore Sifaris Sayi')
plt.xlabel('YEAR_ID')
plt.ylabel('Sifaris Sayi')
plt.show()
```

```python
df_range = df[(df['SALES'] >= 1000) & (df['SALES'] <= 5000)]
countries = df_range['COUNTRY'].unique()
data_by_country = [df_range[df_range['COUNTRY'] == c]['SALES'] for c in countries]
plt.figure(figsize=(10, 6))
plt.hist(data_by_country, bins=20, stacked=True, label=countries, edgecolor='black')

plt.title('Satis paylanmasi - Olkeler uzre')
plt.xlabel('SALES')
plt.ylabel('Tezlik')
plt.legend(fontsize=8, ncol=2)
plt.tight_layout()
plt.show()
```
